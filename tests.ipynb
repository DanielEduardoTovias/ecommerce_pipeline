```python
#   ----- lambda en numeros:
numeros = [1, 2, 3, 4, 5]
resultado = list(map(lambda x: x * 2, numeros))
print(resultado)
```

```python
#   ----- regex:
import re
texto = "hola       mundo" #text con un espacio más
resultado = re.sub(r'\s+', ' ', texto)
print(texto)
print("texto sin espacios: ", resultado)
```

```python
#   ----- lambda con apply:

import pandas as pd
df = pd.DataFrame({"nombre": ["daniel", "pedro", "juan"]})
#pandas toma cada elemento de la columna y la pasa a lambda
df["nombre"] = df["nombre"].apply(lambda x: x.upper())
print(df)
```

```python
#   ----- lambda en texto:
nombres = ["daniel", "pedro", "juan"]
res = list(map(lambda x: x.upper(), nombres))
print(res)
```

```python
#   ----- .strip()
texto = "     Hola mundo     "
print(texto.strip())
```

```python
#regex + .strip()
texto = "     Hola       mundo      Python     "

resultado = re.sub(r'\s+', ' ', texto).strip()

print(resultado)
```
